## 2. Handle Missing Values

```python
df.isnull().sum()
```

```python
print(df.columns)
```

```python
for col in df.columns:
    print(col)
```

```python
df.columns = df.columns.str.strip().str.lower()
```

```python
df['price_usd'].fillna(df['price_usd'].mean(), inplace=True)
df['country'].fillna(df['country'].mode()[0], inplace=True)
```

```python
# Numerical columns
num_cols = ['price_usd', 'units_sold', 'profit_usd', 
            'exchange_rate_to_usd', 'gdp_per_capita', 'population_millions']

for col in num_cols:
    df[col].fillna(df[col].mean(), inplace=True)

# Categorical columns
cat_cols = ['country', 'product', 'category']

for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

## 3. Remove Duplicates

```python
df.drop_duplicates(inplace=True)
```

## 4. Outlier Handling

```python
from scipy import stats

z = stats.zscore(df['price_usd'])
df = df[(z < 3) & (z > -3)]
```

## 5. Feature Engineering

```python
# Profit margin
df['profit_margin'] = df['profit_usd'] / df['price_usd']

# Price category
df['price_category'] = pd.cut(df['price_usd'],
                             bins=[0, 50, 100, 200, 500],
                             labels=['Low', 'Medium', 'High', 'Premium'])
```

```python
df.to_csv("cleaned_pricing_data.csv", index=False)
```
